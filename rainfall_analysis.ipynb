**cleaning the data**

1.   Using the try/except
2.   loops and lists



```python
machakos_raw = ["48", "35", "88", "N/A", "140", " 20 ",
                "9", "12", "", "45", "182", "96"]

months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]


def clean_machakos(raw_values):

    cleaned = []
    bad_positions = []

    for position in range(len(raw_values)):
        value = raw_values[position]
        try:
            number = int(value)
            cleaned.append(number)
        except ValueError:
            cleaned.append(0)
            bad_positions.append(position)


    total = 0
    count = 0
    for position in range(len(cleaned)):
        if position not in bad_positions:
            total = total + cleaned[position]
            count = count + 1


    average = round(total / count)

    imputed_months = []
    for position in bad_positions:
        cleaned[position] = average
        imputed_months.append(months[position])

    print("Average of valid values:", average)
    print("Imputed months:", imputed_months)

    return cleaned


machakos_clean = clean_machakos(machakos_raw)
print("Cleaned data:", machakos_clean)
```

**annual rainfall**

```python
import numpy as np

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

rainfall = np.array([
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],
    [48, 35, 88, 68, 140, 20, 9, 12, 68, 45, 182, 96],
])

annual_totals = rainfall.sum(axis=1)

monthly_averages = rainfall.mean(axis=0)

print("Annual totals (mm):", annual_totals)
print("Monthly averages (mm):", monthly_averages)

print("\nAnnual total per county:")
for i in range(len(counties)):
    print(f"  {counties[i]}: {annual_totals[i]} mm")

print("\nAverage per month across all counties:")
for i in range(len(months)):
    print(f"  {months[i]}: {monthly_averages[i]:.1f} mm")
```

```python
import numpy as np

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]
months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

rainfall = np.array([
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],
    [48, 35, 88, 68, 140, 20, 9, 12, 68, 45, 182, 96],
])

wettest_positions = np.argmax(rainfall, axis=1)
print("Wettest positions:", wettest_positions)

wettest_months = [months[position] for position in wettest_positions]

print("\nWettest month per county:")
for county, month in zip(counties, wettest_months):
    print(f"  {county}: {month}")
```

**highesst consecutive rainfall**

```python
import numpy as np

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]
months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

R = np.array([
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],
    [48, 35, 88, 68, 140, 20, 9, 12, 68, 45, 182, 96],
])

pair_sums = R[:, :-1] + R[:, 1:]
print("Shape of pair_sums:", pair_sums.shape)
print(pair_sums)

best_start = np.argmax(pair_sums, axis=1)
best_total = pair_sums.max(axis=1)

planting_windows = []
for start in best_start:
    window = months[start] + "-" + months[start + 1]
    planting_windows.append(window)

print("\nRecommended planting window:")
for county, window, total in zip(counties, planting_windows, best_total):
    print(f"  {county}: {window} ({total} mm combined)")
```

**rainfall_summary.csv**

```python
import csv
import numpy as np

counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]
months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

R = np.array([
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70],
    [48, 35, 88, 68, 140, 20, 9, 12, 68, 45, 182, 96],
])

annual_totals = R.sum(axis=1)
wettest_positions = np.argmax(R, axis=1)
dry_months = (R < 50).sum(axis=1)
pair_sums = R[:, :-1] + R[:, 1:]
best_start = np.argmax(pair_sums, axis=1)

wettest_months = [months[p] for p in wettest_positions]
planting_windows = [months[s] + "-" + months[s + 1] for s in best_start]

with open("rainfall_summary.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["county", "annual_total_mm", "wettest_month",
                     "dry_months", "planting_window"])
    for county, total, wettest, dry, window in zip(
            counties, annual_totals, wettest_months, dry_months, planting_windows):
        writer.writerow([county, int(total), wettest, int(dry), window])

with open("rainfall_summary.csv") as file:
    print(file.read())
```

## Planting Advisory

For Nairobi, Nakuru, Kisumu and Mombasa, we recommend planting in April-May, when combined rainfall is highest (between 265 mm and 420 mm). Kisumu is the wettest county, with 1371 mm a year and no dry months, so it has the most reliable rainfall. Machakos is the exception: its best window is November-December (278 mm combined), and it has six dry months, so farmers there should plan around the short late-year rains. Machakos is also the driest county overall at 811 mm, so drought-tolerant crops are worth considering. Two Machakos readings (April and September) were invalid and had to be estimated, so its figures are less certain than the other counties'. These results come from one year of data, so they should be checked against more years before the advice is sent out by SMS.

## Data Quality Notes

- The Machakos row arrived as raw text with errors. "N/A" (April) and an empty value (September) could not be converted to numbers.
- Both invalid values were replaced with the rounded average of the 10 valid Machakos readings (68 mm). The reading " 20 " (June) had extra spaces but was valid, so it was kept as 20.
- 2 of 12 Machakos months (about 17%) are estimates, which adds about 136 mm to its annual total of 811 mm.
- The imputed values do not change the Machakos planting window, because April and September are not part of November-December.
- Neither imputed value counts as a dry month, because 68 mm is above the 50 mm threshold.
- The four other counties came in clean, so no changes were made to them.
- The dataset covers a single year, so the wettest months and planting windows may not hold in every year.
